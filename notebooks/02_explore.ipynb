# 02 — Explore the data

Sanity checks, missing values, distributions, and team-name audit against the World Cup 2026 squad.

```python
import sys, pathlib
ROOT = pathlib.Path.cwd().parent
sys.path.insert(0, str(ROOT))
import pandas as pd, numpy as np
import matplotlib.pyplot as plt
from src.wc2026_config import all_teams, NAME_ALIASES

df = pd.read_csv(ROOT / 'data' / 'raw' / 'results.csv', parse_dates=['date'])
df = df.sort_values('date').reset_index(drop=True)
print(df.shape)
df.head()
```

```python
print(df.dtypes)
print('\nmissing:\n', df.isna().sum())
print('\ntournaments (top 15):\n', df.tournament.value_counts().head(15))
```

```python
df['year'] = df.date.dt.year
df.groupby('year').size().plot(figsize=(10,3), title='Matches per year')
plt.tight_layout(); plt.show()
```

```python
df['result'] = np.where(df.home_score > df.away_score, 'H',
                np.where(df.home_score < df.away_score, 'A', 'D'))
print(df.result.value_counts(normalize=True).round(3))
```

```python
# Audit: which WC2026 team names don't appear in the dataset under the FIFA spelling?
names_in_data = set(df.home_team).union(df.away_team)
missing = [t for t in all_teams() if t not in names_in_data]
print('WC2026 names missing from dataset:', missing)
print('\nIf this is non-empty, extend NAME_ALIASES in src/wc2026_config.py.')
```
