# 01 — Download datasets

Downloads the **martj42/international_results** dataset (no API key needed) into `data/raw/`. ~47k international matches since 1872.

Run this notebook first.

```python
import os, sys, pathlib, requests
ROOT = pathlib.Path.cwd().parent
RAW = ROOT / 'data' / 'raw'
RAW.mkdir(parents=True, exist_ok=True)
print('Raw data dir:', RAW)
```

```python
BASE = 'https://raw.githubusercontent.com/martj42/international_results/master/'
FILES = ['results.csv', 'goalscorers.csv', 'shootouts.csv', 'former_names.csv']

for f in FILES:
    url = BASE + f
    dest = RAW / f
    if dest.exists() and dest.stat().st_size > 0:
        print(f'skip (exists): {f}  {dest.stat().st_size:,} bytes')
        continue
    print(f'downloading {url} ...')
    r = requests.get(url, timeout=60)
    r.raise_for_status()
    dest.write_bytes(r.content)
    print(f'  saved {dest} ({len(r.content):,} bytes)')
```

```python
import pandas as pd
df = pd.read_csv(RAW / 'results.csv')
print('rows:', len(df))
print('date range:', df.date.min(), '->', df.date.max())
df.head()
```
